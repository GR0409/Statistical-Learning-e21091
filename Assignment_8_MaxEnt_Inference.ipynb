# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

<br>

**Answers**

<br>

1.
The canonical partition function is $Z(T) = \sum_{j=0}^{\infty} e^{-E_j / (k_B T)}$.

Substitute $E_j = \hbar\omega(j + 1/2)$ and let $\beta = \frac{1}{k_B T}$,

$$Z(T) = \sum_{j=0}^{\infty} e^{-\beta \hbar \omega (j + 1/2)} = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^{\infty} (e^{-\beta \hbar \omega})^j$$

Apply the sum of an infinite geometric series $\sum_{j=0}^{\infty} r^j = \frac{1}{1-r}$ where $r = e^{-\beta \hbar \omega}$,

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

2.
Helmholtz Free Energy, $F(T) = -k_B T \ln Z$,

$$F(T) = -k_B T \left( -\frac{\hbar \omega}{2k_BT} - \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) \right)$$

$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right)$$

Internal Energy, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$,

$$\ln Z = -\frac{\beta \hbar \omega}{2} - \ln(1 - e^{-\beta \hbar \omega})$$

$$U(T) = -\left( -\frac{\hbar \omega}{2} - \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}} \right)$$

$$U(T) = \hbar \omega \left( \frac{1}{2} + \frac{1}{e^{\frac{\hbar \omega}{k_BT}} - 1} \right)$$

3. Using $S = -\left(\frac{\partial F}{\partial T}\right)$,

$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) - k_B T \left( \frac{1}{1 - e^{-\frac{\hbar \omega}{k_BT}}} \right) \left(- e^{-\frac{\hbar \omega}{k_BT}}\right) \left(\frac{\hbar \omega}{k_B T^2}\right)$$

$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) + \frac{\hbar \omega}{T} \frac{1}{e^{\frac{\hbar \omega}{k_BT}} - 1}$$

Verification of $U = F + TS$,

$$F + TS = \frac{\hbar \omega}{2} + k_B T \ln(1 - e^{-\beta \hbar \omega}) - k_B T \ln(1 - e^{-\beta \hbar \omega}) + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1}$$

$$F + TS = \hbar \omega \left(\frac{1}{2} + \frac{1}{e^{\beta \hbar \omega} - 1}\right) = U$$

4. Heat Capacity, $C_V = \left(\frac{\partial U}{\partial T}\right)$,

$$C_V = \frac{\partial}{\partial T} \left( \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_BT}} - 1} \right) = \hbar \omega \left( - \frac{1}{(e^{\frac{\hbar \omega}{k_BT}} - 1)^2} \right) e^{\frac{\hbar \omega}{k_BT}} \left( - \frac{\hbar \omega}{k_B T^2} \right)$$

$$C_V = k_B \left( \frac{\hbar \omega}{k_B T} \right)^2 \frac{e^{\frac{\hbar \omega}{k_BT}}}{(e^{\frac{\hbar \omega}{k_BT}} - 1)^2}$$

Hyperbolic Substitution,

Using $\sinh(x) = \frac{e^x - e^{-x}}{2}$, we square and rearrange it,

$\frac{1}{\sinh^2(x)} = \frac{4}{(e^x - e^{-x})^2} = \frac{4e^{2x}}{(e^{2x} - 1)^2}$

Let $x = \frac{\hbar \omega}{2k_BT}$, meaning $2x = \frac{\hbar \omega}{k_BT}$

Substituting this into the $C_V$ fraction,

$$\frac{e^{\frac{\hbar \omega}{k_BT}}}{(e^{\frac{\hbar \omega}{k_BT}} - 1)^2} = \frac{1}{4\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$

$$C_V = k_B \left(\frac{\hbar \omega}{k_BT}\right)^2 \frac{1}{4\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)} = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$

5. Asymptotic BehaviorIn the high-temperature limit $k_B T \gg \hbar \omega$, the term $x = \frac{\hbar \omega}{2k_BT}$ becomes very small ($x \ll 1$).For small $x$, $\sinh(x) \approx x$.Therefore, $C_V \approx k_B (x)^2 \frac{1}{(x)^2} = k_B$.

Conclusion - Yes, this matches the classical equipartition theorem prediction of $k_B$ per oscillator contributions of $\frac{1}{2}k_B$ for kinetic and $\frac{1}{2}k_B$ for potential energy.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

<br>

**Answers**

<br>

1. Maximize Shannon Entropy $H = -\int p(x) \ln p(x) dx$ subject to constraints.

$$\mathcal{L} = -\int p(x) \ln p(x) dx + \lambda_0 \left(1 - \int p(x) dx\right) + \lambda_1 \left(\mu - \int x p(x) dx\right) + \lambda_2 \left(\sigma^2 - \int (x-\mu)^2 p(x) dx\right)$$

2. Take the functional derivative with respect to $p(x)$ and set to zero,

$$\frac{\delta \mathcal{L}}{\delta p(x)} = -\ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0$$

$$p(x) = \exp(-1 - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2)$$

Because the distribution must be symmetric around the mean $\mu$ (due to the variance constraint purely capturing dispersion around $\mu$), $\lambda_1 = 0$.

Let $A = e^{-1-\lambda_0}$.

$$p(x) = A \exp(-\lambda_2 (x-\mu)^2)$$

3. Normalization constraint ($\int p(x) dx = 1$):$$A \int_{-\infty}^{\infty} \exp(-\lambda_2 (x-\mu)^2) dx = A \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies A = \sqrt{\frac{\lambda_2}{\pi}}$$

Variance constraint ($\int (x-\mu)^2 p(x) dx = \sigma^2$),

$$\int_{-\infty}^{\infty} (x-\mu)^2 A \exp(-\lambda_2 (x-\mu)^2) dx = A \left( \frac{1}{2\lambda_2} \sqrt{\frac{\pi}{\lambda_2}} \right) = \sigma^2$$

Substitute $A = \sqrt{\frac{\lambda_2}{\pi}}$,

$$\sqrt{\frac{\lambda_2}{\pi}} \left( \frac{1}{2\lambda_2} \sqrt{\frac{\pi}{\lambda_2}} \right) = \frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$

Final Substitution,

$$A = \sqrt{\frac{1}{2\pi\sigma^2}} = \frac{1}{\sqrt{2\pi\sigma^2}}$$

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




<br>

**Answers**

<br>

Partition Function,

$$Z(T) = \sum_{j=0}^{\infty} e^{-\beta \hbar \omega (j + 1/2)} = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^{\infty} (e^{-\beta \hbar \omega})^j = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

Thermodynamic Potentials,

$$F(T) = -k_B T \ln Z = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right)$$

$$U(T) = -\frac{\partial \ln Z}{\partial \beta} = \hbar \omega \left( \frac{1}{2} + \frac{1}{e^{\frac{\hbar \omega}{k_BT}} - 1} \right)$$

Entropy,

$$S(T) = -\frac{\partial F}{\partial T} = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_BT}}\right) + \frac{\hbar \omega}{T} \frac{1}{e^{\frac{\hbar \omega}{k_BT}} - 1}$$

Heat Capacity,

$$C_V = \frac{\partial U}{\partial T} = k_B \left( \frac{\hbar \omega}{k_B T} \right)^2 \frac{e^{\frac{\hbar \omega}{k_BT}}}{(e^{\frac{\hbar \omega}{k_BT}} - 1)^2}$$

Substitute $\sinh(x) = \frac{e^x - e^{-x}}{2}$ where $x = \frac{\hbar \omega}{2k_BT}$,

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$

Limit - As $T \to \infty$, $x \to 0$, $\sinh(x) \approx x \implies C_V \to k_B$ (Matches equipartition).

2. MaxEnt Lagrangian,

$$\mathcal{L} = -\int p \ln p dx - \lambda_0\left(1 - \int p dx\right) - \lambda_1\left(\mu - \int x p dx\right) - \lambda_2\left(\sigma^2 - \int (x-\mu)^2 p dx\right)$$

Optimization,

$$\frac{\delta \mathcal{L}}{\delta p} = -\ln p - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0$$

$$p(x) = \exp(-1 - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2)$$

Constraints applied,

Symmetry around mean dictates $\lambda_1 = 0$.Normalization dictates $e^{-1-\lambda_0} = \sqrt{\frac{\lambda_2}{\pi}}$.

Variance dictates $\lambda_2 = \frac{1}{2\sigma^2}$.

$$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

3. Multiclass Logistic Regression Model,

$$P(Y=1|\mathbf{x}) = \sigma(\lambda_1 x_1 + \lambda_2 x_2 + \lambda_3 x_3)$$

System of Equations,

$$\mathbb{E}[x_i \cdot \mathbb{I}(Y=1)] = c_i$$

$0.40\sigma(\lambda_1) + 0.05\sigma(\lambda_1+\lambda_2) = 0.05$

$0.25\sigma(\lambda_2) + 0.15\sigma(\lambda_2+\lambda_3) + 0.05\sigma(\lambda_1+\lambda_2) = 0.15$

$0.05\sigma(\lambda_3) + 0.15\sigma(\lambda_2+\lambda_3) = 0.16$

Probabilities,

$p_A \approx 0.122$, $p_B \approx 0.145$, $p_C \approx 0.947$, $p_D \approx 0.752$, $p_E \approx 0.023$

Final Weights,

$$\lambda_i = \ln\left(\frac{p_i}{1-p_i}\right)$$

$\lambda_1 \approx -1.97$$\lambda_2 \approx -1.77$$\lambda_3 \approx 2.88$

4. Asset PricingLagrangian Optimization,

$$\frac{\delta \mathcal{L}}{\delta p(S_T)} = -\ln p(S_T) - 1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0$$

$$p(S_T) = \exp\left(-1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)\right)$$

$$p(S_T) = C \cdot S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$

(where $C = e^{-1-\lambda_0}$)



# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



<br>

**Answers**

<br>

1. Normalization,

$\int_0^\infty p(S_T) dS_T = 1$

Expected Price,

$\int_0^\infty S_T p(S_T) dS_T = 102$

Expected Squared Price,

$\int_0^\infty S_T^2 p(S_T) dS_T = 10600$

Logarithmic Constraint,

$\int_0^\infty \ln(S_T) p(S_T) dS_T = 4.60$2.

Formulate the Lagrangian,

$$\mathcal{L} = -\int_0^\infty p(S_T) \ln p(S_T) dS_T - \lambda_0 \left(1 - \int_0^\infty p(S_T) dS_T\right) - \lambda_1 \left(102 - \int_0^\infty S_T p(S_T) dS_T\right) - \lambda_2 \left(10600 - \int_0^\infty S_T^2 p(S_T) dS_T\right) - \lambda_3 \left(4.60 - \int_0^\infty \ln(S_T) p(S_T) dS_T\right)$$

3. Functional Optimization,

Set the functional derivative with respect to $p(S_T)$ to zero,

$$\frac{\delta \mathcal{L}}{\delta p(S_T)} = -\ln p(S_T) - 1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0$$

4. Solve for the Probability Density Function $p(S_T)$,

$$\ln p(S_T) = -1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)$$

$$p(S_T) = \exp\left(-1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)\right)$$

5. Factor out the constants and the logarithmic term (using $e^{-\lambda_3 \ln(S_T)} = S_T^{-\lambda_3}$), and let $C = e^{-1-\lambda_0}$,

$$p(S_T) = C \cdot S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$

$$C \int_0^\infty S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) dS_T = 1$$

$$C \int_0^\infty S_T^{(1-\lambda_3)} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) dS_T = 102$$

$$C \int_0^\infty S_T^{(2-\lambda_3)} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) dS_T = 10600$$

$$C \int_0^\infty \ln(S_T) S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) dS_T = 4.60$$
