#Q3) Transformer Failiures

In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$
<br><br>
1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

$\Omega = \{0, 1, 2, 3\}$

$\mathfrak{F} = \mathfrak{P}(\Omega)$

$P(\{0\}) = 0.50$

$P(\{1\}) = 0.30$

$P(\{2\}) = 0.15$

$P(\{3\}) = 0.05$

$A \in \mathfrak{F}$

$P(A) = \sum_{\omega \in A} P(\{\omega\})$

<br>
2. Define $X$ as a random variable on this space.

$X(\omega) = \omega \quad \text{for all } \omega \in \Omega$

<br>
3. Determine the probability mass function of $X$.

$p_X(0) = 0.50$

$p_X(1) = 0.30$

$p_X(2) = 0.15$

$p_X(3) = 0.05$

$p_X(x) = 0$

<br>
4. Compute the expectation $E[X]$.

$$E[X] = \sum_{x=0}^{3} x \cdot P(X=x)$$

$$E[X] = (0 \cdot 0.50) + (1 \cdot 0.30) + (2 \cdot 0.15) + (3 \cdot 0.05)$$

$$E[X] = 0 + 0.30 + 0.30 + 0.15 = \mathbf{0.75}$$

<br>
5. Compute the variance $\mathrm{Var}(X)$.

$$E[X^2] = \sum_{x=0}^{3} x^2 \cdot P(X=x)$$

$$E[X^2] = (0^2 \cdot 0.50) + (1^2 \cdot 0.30) + (2^2 \cdot 0.15) + (3^2 \cdot 0.05)$$

$$E[X^2] = 0 + 0.30 + 0.60 + 0.45 = 1.35$$

$\text{Var}(X) = E[X^2] - (E[X])^2$:

$$\text{Var}(X) = 1.35 - (0.75)^2$$

$$\text{Var}(X) = 1.35 - 0.5625 = \mathbf{0.7875}$$

<br>
6. Find $P(X\ge 1)$ and $P(X\ge 2)$.

$P(X \ge 1) = P(X=1) + P(X=2) + P(X=3) = 0.30 + 0.15 + 0.05 = \mathbf{0.50}$

$1 - P(X=0) = 1 - 0.50 = 0.50$

$P(X \ge 2) = P(X=2) + P(X=3) = 0.15 + 0.05 = \mathbf{0.20}$

<br>
7. Briefly explain the practical meaning of the expectation and variance in this setting.

Expectation ($0.75$) - On an average day, the district can expect 0.75 transformer failures.

Variance ($0.7875$) - This measures the day to day fluctuation or volatility in failures.



# Q4) Probability of Defective Items

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},
\end{cases}
$$

where $B(2,18)$ is the Beta function.

Answer the following:



1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

$\Omega = [0, 1]$

$\mathfrak{F} = \sigma(\{(a, b) \mid 0 \le a < b \le 1\})$

$$f(x) = \frac{1}{B(2,18)} \cdot x^1 \cdot (1 - x)^{17} > 0$$

$P(A) = \int_A \frac{1}{B(2,18)} x^1(1 - x)^{17} dx$

<br>
2. Define $X$ as a continuous random variable on this space.

$X: \Omega \rightarrow \mathbb{R}$

$X(\omega) = \omega, \quad \omega \in [0, 1]$

$$f(x) = \begin{cases} \frac{1}{B(2,18)} x (1 - x)^{17}, & 0 < x < 1 \\ 0, & \text{otherwise} \end{cases}$$

<br>
4. Verify that $f$ is a valid probability density function.

For $0 < x < 1$, both $x > 0$ and $(1-x)^{17} > 0$.

Since the Beta function $B(2, 18)$ is a positive constant, $f(x) \ge 0$ for all $x$.Integrates to 1:

$$\int_{-\infty}^{\infty} f(x) \, dx = \int_{0}^{1} \frac{1}{B(2,18)} x^{2-1} (1-x)^{18-1} \, dx$$

$\int_0^1 x^{\alpha-1}(1-x)^{\beta-1}dx = B(\alpha, \beta)$.

Therefore:$$\frac{1}{B(2,18)} \int_{0}^{1} x^{2-1} (1-x)^{18-1} \, dx = \frac{1}{B(2,18)} \cdot B(2,18) = 1$$

<br>
5. Compute the expectation $E[X]$.

$$E[X] = \frac{2}{2 + 18} = \frac{2}{20} = \mathbf{0.10}$$

<br>
6. Compute the variance $\mathrm{Var}(X)$.

$\text{Var}(X) = \frac{\alpha\beta}{(\alpha + \beta)^2(\alpha + \beta + 1)}$

$$\text{Var}(X) = \frac{2 \cdot 18}{(2 + 18)^2(2 + 18 + 1)}$$

$$\text{Var}(X) = \frac{36}{(20)^2(21)} = \frac{36}{400 \cdot 21} = \frac{36}{8400} = \mathbf{\frac{3}{700}} \approx \mathbf{0.00429}$$

# Q5) Expected Project Completion Time

A construction company is estimating the time required to install a backup generator at a hospital. Let $X$ denote the number of hours needed to complete the installation. Based on expert judgment, the shortest possible time is $4$ hours, the longest possible time is $10$ hours, and the most likely completion time is $6$ hours. Assume that $X$ follows a Triangular distribution with parameters $a=4$, $b=10$, and $c=6$.

Answer the following:




1. Specify the support of $X$.

$X = [4, 10]$

<br>
2. Write down the probability density function $f_X(x)$.

$$f_X(x) = \begin{cases} \frac{2(x-4)}{(10-4)(6-4)}, & 4 \le x \le 6 \\ \frac{2(10-x)}{(10-4)(10-6)}, & 6 < x \le 10 \\ 0, & \text{otherwise} \end{cases}$$

$$f_X(x) = \begin{cases} \frac{x-4}{6}, & 4 \le x \le 6 \\ \frac{10-x}{12}, & 6 < x \le 10 \\ 0, & \text{otherwise} \end{cases}$$

<br>
3. Verify that $f_X$ integrates to $1$ over its support.

$$\int_{4}^{10} f_X(x) dx = \int_{4}^{6} \frac{x-4}{6} dx + \int_{6}^{10} \frac{10-x}{12} dx$$

$$= \frac{1}{6} \left[ \frac{x^2}{2} - 4x \right]_4^6 + \frac{1}{12} \left[ 10x - \frac{x^2}{2} \right]_6^{10}$$

$$= \frac{1}{6} \left[ \left(\frac{36}{2} - 24\right) - \left(\frac{16}{2} - 16\right) \right] + \frac{1}{12} \left[ \left(100 - \frac{100}{2}\right) - \left(60 - \frac{36}{2}\right) \right]$$

$$= \frac{1}{6} \left[ (18 - 24) - (8 - 16) \right] + \frac{1}{12} \left[ 50 - (60 - 18) \right]$$

$$= \frac{1}{6} \left[ -6 - (-8) \right] + \frac{1}{12} \left[ 50 - 42 \right]$$

$$= \frac{1}{6} [2] + \frac{1}{12} [8]$$

$$= \frac{1}{3} + \frac{2}{3} = \mathbf{1}$$

<br>
4. Compute the expectation $E[X]$.

$$E[X] = \frac{a + b + c}{3}$$

$$E[X] = \frac{4 + 10 + 6}{3}$$

$$E[X] = \mathbf{\frac{20}{3}} \approx \mathbf{6.67}$$

<br>
5. Compute the variance $\mathrm{Var}(X)$.

$$\text{Var}(X) = \frac{a^2 + b^2 + c^2 - ab - ac - bc}{18}$$

$$\text{Var}(X) = \frac{4^2 + 10^2 + 6^2 - (4)(10) - (4)(6) - (10)(6)}{18}$$

$$\text{Var}(X) = \frac{16 + 100 + 36 - 40 - 24 - 60}{18}$$

$$\text{Var}(X) = \frac{152 - 124}{18}$$

$$\text{Var}(X) = \mathbf{\frac{28}{18}} = \mathbf{\frac{14}{9}} \approx \mathbf{1.56}$$

<br>
6. Interpret the values of $a$, $b$, and $c$ in the context of this problem.

$a = 4$: The optimistic or minimum possible time to complete the installation.

$b = 10$: The pessimistic or maximum possible time to complete the installation.

$c = 6$: The most likely time to complete the installation.

<br>
7. Explain briefly why a Triangular distribution is a reasonable model in this setting.

The Triangular distribution is highly useful in project management because it relies entirely on intuitive, expert-driven estimates the best-case, worst-case, most likely when exact historical data is unavailable.

8. Compute the probability that the installation takes at most $5$ hours, that is, compute

$$
P(X \leq 5).
$$

$$P(X \le 5) = \int_{4}^{5} \frac{x-4}{6} dx$$

$$P(X \le 5) = \frac{1}{6} \left[ \frac{x^2}{2} - 4x \right]_4^5$$

$$P(X \le 5) = \frac{1}{6} \left[ \left(\frac{25}{2} - 20\right) - \left(\frac{16}{2} - 16\right) \right]$$

$$P(X \le 5) = \frac{1}{6} \left[ -7.5 - (-8) \right]$$

$$P(X \le 5) = \frac{1}{6} [0.5] = \mathbf{\frac{1}{12}}$$

<br>
9. Compute the probability that the installation takes between $5$ and $8$ hours, that is, compute

$$
P(5 \leq X \leq 8).
$$

$$P(5 \le X \le 8) = \int_{5}^{6} \frac{x-4}{6} dx + \int_{6}^{8} \frac{10-x}{12} dx$$


$$\int_{5}^{6} \frac{x-4}{6} dx = \int_{4}^{6} \frac{x-4}{6} dx - \int_{4}^{5} \frac{x-4}{6} dx$$

$$= \frac{1}{3} - \frac{1}{12} = \frac{3}{12} = \frac{1}{4}$$

$$\int_{6}^{8} \frac{10-x}{12} dx = \frac{1}{12} \left[ 10x - \frac{x^2}{2} \right]_6^8$$

$$= \frac{1}{12} \left[ \left(80 - \frac{64}{2}\right) - \left(60 - \frac{36}{2}\right) \right]$$

$$= \frac{1}{12} \left[ (80 - 32) - (60 - 18) \right]$$

$$= \frac{1}{12} \left[ 48 - 42 \right] = \frac{6}{12} = \frac{1}{2}$$

Sum:

$$P(5 \le X \le 8) = \frac{1}{4} + \frac{1}{2} = \mathbf{\frac{3}{4}}$$

#Q6) Expected Project Completion Time Using PERT

A project manager is estimating the time required to complete a critical software integration task. Let $X$ denote the number of days required to complete the task. Based on expert judgment, the shortest possible time is $a=8$ days, the longest possible time is $b=20$ days, and the most likely completion time is $c=11$ days. Assume that $X$ follows a PERT distribution with parameters $(a,b,c)$.

The PERT distribution is defined as a Beta distribution scaled to the interval $[a,b]$, with internal shape parameters

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right).
$$

Its probability density function is

$$
f(x)=\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}}, \qquad a \leq x \leq b,
$$

where

$$
B(\alpha,\beta)=\int_0^1 t^{\alpha-1}(1-t)^{\beta-1},dt.
$$

Answer the following:

<br>



1. Compute the values of the shape parameters $\alpha$ and $\beta$.

$$\alpha = 1 + 4\left(\frac{c - a}{b - a}\right) = 1 + 4\left(\frac{11 - 8}{20 - 8}\right) = 1 + 4\left(\frac{3}{12}\right) = 1 + 1 = \mathbf{2}$$

$$\beta = 1 + 4\left(\frac{b - c}{b - a}\right) = 1 + 4\left(\frac{20 - 11}{20 - 8}\right) = 1 + 4\left(\frac{9}{12}\right) = 1 + 3 = \mathbf{4}$$

<br>
2. Specify the support of $X$.

$X = [a, b] = \mathbf{[8, 20]}$

<br>
3. Write down the probability density function $f_X(x)$ for this model.

$$f_X(x) = \frac{(x - a)^{\alpha - 1}(b - x)^{\beta - 1}}{B(\alpha, \beta)(b - a)^{\alpha + \beta - 1}}$$

$$f_X(x) = \frac{(x - 8)^{2 - 1}(20 - x)^{4 - 1}}{B(2, 4)(20 - 8)^{2 + 4 - 1}}$$

$$f_X(x) = \begin{cases} \mathbf{\frac{(x - 8)(20 - x)^3}{B(2, 4) \cdot 12^5}}, & \mathbf{8 \le x \le 20} \\ 0, & \text{otherwise} \end{cases}$$

<br>
4. State the probability space $(\Omega,\mathfrak F,P)$ suitable for this continuous model.

$\Omega = [8, 20]$

$\mathfrak{F} = \mathfrak{B}([8, 20])$

$P(A) = \int_A f_X(x) \, dx$

<br>
5. Define $X$ as a random variable on this space.

$X: \Omega \rightarrow \mathbb{R}$

$X(\omega) = \omega, \quad \omega \in [8, 20]$

<br>
6. Show that the expectation $E[X]$ is given by

$$
E[X]=\frac{a+4c+b}{6}.
$$

$$E[X] = \frac{a + 4c + b}{6}$$

$$E[X] = \frac{8 + 4(11) + 20}{6}$$

$$E[X] = \frac{8 + 44 + 20}{6} = \frac{72}{6} = \mathbf{12}$$

<br>
7. Compute the variance $\mathrm{Var}(X)$ using the approximation

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}.
$$

$$\text{Var}(X) = \frac{(E[X] - a)(b - E[X])}{7}$$

$$\text{Var}(X) = \frac{(12 - 8)(20 - 12)}{7}$$

$$\text{Var}(X) = \frac{(4)(8)}{7} = \mathbf{\frac{32}{7}} \approx \mathbf{4.57}$$

<br>
8. Compute the approximate standard deviation $\sigma$.

$$\sigma = \sqrt{\text{Var}(X)}$$

$$\sigma = \sqrt{\frac{32}{7}} \approx \mathbf{2.14}$$

<br>
9. Briefly explain why a PERT distribution may be more appropriate than a Triangular distribution in this setting byt comparing the two results.

The PERT distribution relies on a smooth bell-shaped curve that concentrates more probability around the most likely value and places less weight on the extreme tails. This makes it a more realistic model for human time-estimates than the straight, rigid lines of a Triangular distribution.

<br>
10. Interpret the practical meaning of the values $a$, $b$, $c$, $E[X]$, and $\sigma$ in the context of project planning.

$a = 8$ - The absolute minimum days to finish the integration.

$b = 20$ - The absolute maximum days to finish if everything goes wrong.

$c = 11$ - The most likely days to finish.

$E[X] = 12$ - The expected average completion time for the project, factoring in the risk of delays.

$\sigma \approx 2.14$ - The standard deviation, representing the uncertainty or spread of the project's timeline around the expected 12 days.
